```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def step_fun(x):
    return np.where(x>0,1,0)
```

```python
x=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])
```

```python
y=np.array([0,0,0,1])
```

```python
weights= np.array([1.4, 0.5])
bias=0
learning_rate=0.5
epochs=10
```

```python
for epoch in range(epochs):
    print(f"\nEpoch {epoch+1}")
    errors=0

    for i in range(len(x)):
        linear_output=np.dot(x[i],weights)+bias
        prediction=step_fun(linear_output)

        error=y[i]-prediction

        weights+=learning_rate*error*x[i]
        bias+=learning_rate*error

        if error!=0:
            errors+=1
        print(f"Input: {x[i]},Target: {y[i]}, Prediction: {prediction},"
        f"Weights: {weights},Bias :{bias}") 
    if errors==0:
        print("\n Traning Converaged Successfully")
        break
    
```

```python
print("\nFinal Weights: ",weights)
print("Final ")
```

```python
print("\n Testing the perceptions")
for sample in x:
    output=step_fun(np.dot(sample,weights)+bias)
    print(f"Input: {sample}-->output: {output}")
                    
```
